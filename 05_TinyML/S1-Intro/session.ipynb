```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import os
```

```python
# Import required libraries for downloading and extracting
import urllib.request
import zipfile
import os

# Dataset URL
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/00240/UCI%20HAR%20Dataset.zip"

# File name to save
zip_path = "har_dataset.zip"

# Download the dataset
urllib.request.urlretrieve(url, zip_path)

print("Download complete.")

# Extract the dataset
with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall("har_data")

print("Extraction complete.")
```

```python
x_train = pd.read_csv("har_data/UCI HAR Dataset/train/X_train.txt", delim_whitespace=True, header=None)  # No need to exclude columns, this is pure features
y_train = pd.read_csv("har_data/UCI HAR Dataset/train/y_train.txt", delim_whitespace=True, header=None).iloc[:,0] - 1  # Get labels and 0-index them
x_test = pd.read_csv("har_data/UCI HAR Dataset/test/X_test.txt", delim_whitespace=True, header=None)
y_test = pd.read_csv("har_data/UCI HAR Dataset/test/y_test.txt", delim_whitespace=True, header=None).iloc[:,0] - 1
```

```python
x_train
```

```python
# need to apply feature selection to reduce the number of features to 10-20 features. This will help in reducing the model size and also improve the performance of the model on embedded devices.

#1. SCALING THE DATA
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
# replaces to feature selection code to select the first 20 features. This is a simple approach to reduce the number of features and it will help in reducing the model size and also improve the performance of the model on embedded devices.
x_train = x_train_scaled[:,:20]  # Select the first 20 features
x_test = x_test_scaled[:,:20]    # Select the first 20 features
```

```python
x_train.shape, y_train.shape, x_test.shape, y_test.shape
```

```python
model=keras.Sequential([
    layers.Input(shape=(20,)),
    layers.Dense(64, activation='relu'),
    layers.Dense(32, activation='relu'),
    layers.Dense(6, activation='softmax')
])  
```

```python
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
model.summary()
```

```python
model.fit(x_train, y_train, epochs=100, validation_data=(x_test, y_test))
```

```python
#iniate converter
converter = tf.lite.TFLiteConverter.from_keras_model(model)
tflite_model = converter.convert()
with open("model.tflite", "wb") as f:
    f.write(converter.convert())
```

```python
# quantization
converter = tf.lite.TFLiteConverter.from_keras_model(model)
converter.optimizations = [tf.lite.Optimize.DEFAULT]
tflite_quant_model = converter.convert()
with open("model_quant.tflite", "wb") as f:
    f.write(tflite_quant_model)
```

```python
print("original model size: {:.2f} KB".format(os.path.getsize("model.tflite") / 1024))
print("quantized model size: {:.2f} KB".format(os.path.getsize("model_quant.tflite") / 1024))
```

```python
interpreter = tf.lite.Interpreter(model_path="model_quant.tflite")
interpreter.allocate_tensors()
```

```python
iput_details = interpreter.get_input_details()
output_details = interpreter.get_output_details()
```

```python
sample=np.array([x_test[0]], dtype=np.float32)
interpreter.set_tensor(iput_details[0]['index'], sample)
interpreter.invoke()
output = interpreter.get_tensor(output_details[0]['index'])
print("Predicted class:", np.argmax(output))
```
