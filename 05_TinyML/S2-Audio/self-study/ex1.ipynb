```python
import numpy as np

def frame_audio(signal, sample_rate=16000, frame_ms=25, hop_ms=10):
    """
    تقسيم الصوت إلى frames
    
    Parameters:
    -----------
    signal : array - الصوت الخام
    sample_rate : int - معدل العينات (16000 Hz)
    frame_ms : int - طول الـ frame بالمللي ثانية
    hop_ms : int - خطوة التقدم بالمللي ثانية
    """
    
    # حساب المعاملات بالعينات
    frame_length = int(sample_rate * frame_ms / 1000)  # 400 عينة
    hop_length = int(sample_rate * hop_ms / 1000)      # 160 عينة
    
    # عدد الـ frames
    num_frames = 1 + (len(signal) - frame_length) // hop_length
    
    # إنشاء الـ frames
    frames = []
    for i in range(num_frames):
        start = i * hop_length
        end = start + frame_length
        frame = signal[start:end]
        frames.append(frame)
    
    return np.array(frames), frame_length, hop_length

# مثال
# ═══════════════════════════════════════════════════════════════
# signal = [16000 عينة]  (ثانية واحدة)
# 
# النتيجة:
# ├── عدد الـ frames ≈ 98 frame
# ├── كل frame = 400 عينة
# └── شكل المصفوفة: (98, 400)
```

```python
import numpy as np

def compute_fft(frame, n_fft=512):
    """
    حساب FFT لـ frame واحد
    """
    # إضافة padding لجعل الطول قوة 2
    if len(frame) < n_fft:
        frame = np.pad(frame, (0, n_fft - len(frame)))
    
    # حساب FFT
    fft_result = np.fft.fft(frame, n=n_fft)
    
    # أخذ النصف الأول + حساب القيمة المطلقة
    magnitude = np.abs(fft_result[:n_fft//2 + 1])
    
    return magnitude

# مثال
# ═══════════════════════════════════════════════════════════════
frame = np.random.randn(400)  # frame واحد
spectrum = compute_fft(frame, n_fft=512)

print(f"شكل المدخل: {frame.shape}")      # (400,)
print(f"شكل المخرج: {spectrum.shape}")   # (257,)

# الترددات المقابلة
frequencies = np.linspace(0, 8000, 257)  # من 0 إلى Nyquist
```
