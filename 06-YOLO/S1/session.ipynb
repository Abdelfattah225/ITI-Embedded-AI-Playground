```python
from ultralytics import YOLO
```

```python
# Define the path to the dataset YAML file
DATA_YAML = "coco128.yaml"
# Load the YOLOv8 model (you can specify the version, e.g., 'yolov8n.pt' for nano)
model = YOLO('yolov8n.pt') # model lightwight

# add 
model.train(data=DATA_YAML, epochs=10, imgsz=640, batch=16,patience=5, name='yolov8n_coco128')



```

```python
best_model_path = 'runs/detect/yolov8n_coco128/weights/best.pt'
best = YOLO(best_model_path)
```

```python
results = best.val(data=DATA_YAML)
print(results)
```

```python
import cv2
import matplotlib.pyplot as plt
import pathlib
import os
```

```python
pred=best.predict(source='https://ultralytics.com/images/zidane.jpg', save=True, save_txt=True, conf=0.5)
```

```python
save_dir = pathlib.Path(pred[0].save_dir)
out_img=next(save_dir.glob('zidane.jpg'))
img = cv2.cvtColor(cv2.imread(out_img), cv2.COLOR_BGR2RGB)
plt.imshow(img)
plt.axis('off')
plt.show()
```
