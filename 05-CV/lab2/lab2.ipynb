```python
# Import necessary libraries
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from tensorflow import keras
from tensorflow.keras import layers, models
from tensorflow.keras.datasets import cifar10
import random

# Set random seeds for reproducibility
np.random.seed(42)
random.seed(42)

# Load CIFAR-10 dataset
print("Loading CIFAR-10 dataset...")
(x_train_cifar, y_train_cifar), (x_test_cifar, y_test_cifar) = cifar10.load_data()

# Normalize pixel values to [0, 1]
x_train_cifar = x_train_cifar.astype('float32') / 255.0
x_test_cifar = x_test_cifar.astype('float32') / 255.0

# Flatten labels
y_train_cifar = y_train_cifar.flatten()
y_test_cifar = y_test_cifar.flatten()

# Define CIFAR-10 class names
class_names = ['Airplane', 'Automobile', 'Bird', 'Cat', 'Deer', 
               'Dog', 'Frog', 'Horse', 'Ship', 'Truck']

# Print dataset information
print(f"\nCIFAR-10 Training samples: {x_train_cifar.shape[0]}")
print(f"CIFAR-10 Test samples: {x_test_cifar.shape[0]}")
print(f"Image shape: {x_train_cifar.shape[1:]}")
print(f"Number of classes: {len(class_names)}")
print(f"Classes: {class_names}")

# Visualize sample CIFAR-10 images
fig, axes = plt.subplots(2, 5, figsize=(12, 5))
fig.suptitle('Sample CIFAR-10 Images', fontsize=16)
for i, ax in enumerate(axes.flat):
    ax.imshow(x_train_cifar[i])
    ax.set_title(class_names[y_train_cifar[i]])
    ax.axis('off')
plt.tight_layout()
plt.show()

print("\n✓ Step 1 Complete: Environment setup successful!")
```
