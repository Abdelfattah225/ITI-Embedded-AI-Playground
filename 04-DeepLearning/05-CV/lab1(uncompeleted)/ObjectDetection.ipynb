```python
import tensorflow as tf 
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Normalization
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()
x_train = x_train.astype("float32") / 255
x_test = x_test.astype("float32") /255
```

```python
# set canvas at 75 * 75 
CANVAS=75


OBJ=28
```

```python
def make_localization_image(minist_img):
    # Create an empty 75x75 grayscale canvas filled with zeros (black background)
    canvas=np.zeros((CANVAS, CANVAS), dtype=np.float32)

    # Generate a random x-coordinate (0 to 47) for placing the 28x28 object
    x_offset=np.random.randint(0, CANVAS-OBJ)
    # Generate a random y-coordinate (0 to 47) for placing the 28x28 object
    y_offset=np.random.randint(0, CANVAS-OBJ)

    # Calculate the bottom-right x-coordinate of the bounding box (x_offset + 28)
    second_corr_x=x_offset+OBJ
    # Calculate the bottom-right y-coordinate of the bounding box (y_offset + 28)
    second_corr_y=y_offset+OBJ
    # Place the MNIST digit (minist_img) onto the canvas at the random position
    canvas[x_offset:x_offset+OBJ, y_offset:y_offset+OBJ] = minist_img
    # Return the canvas with the placed object and its top-left corner coordinates
    bbox=np.array([x_offset/CANVAS, y_offset/CANVAS, second_corr_x/CANVAS, second_corr_y/CANVAS])
    canvas= np.expand_dims(canvas, axis=-1)
    
    return canvas, bbox
```

```python
n_train = 12000
n_test = 2000
train_images = []
train_labels = []
test_images = []
test_labels = []
for i in range(n_train):
    img, label = make_localization_image(x_train[i])
    train_images.append(img)
    train_labels.append(label)
for i in range(n_test):
    img, label = make_localization_image(x_test[i])
    test_images.append(img)
    test_labels.append(label)
```
